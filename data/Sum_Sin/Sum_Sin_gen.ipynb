```python
import numpy as np

# -------------------------------
# Parameter setup
# -------------------------------
n_samples = 10000
n_points = 1024
k_max = 4

# -------------------------------
# Data generation
# -------------------------------
x = np.linspace(0, 1, n_points)
k = np.arange(k_max + 1)

sin_basis = np.sin(2*np.pi*np.outer(k, x))
f_sin_basis = np.sin(2*np.pi*np.outer(k**4, x))

eta = np.random.randn(n_samples, k_max + 1)

g_values = eta @ sin_basis
f_values = eta @ f_sin_basis

# -------------------------------
# Save dataset
# -------------------------------
np.savez('test_data.npz', input=g_values, output=f_values)
print("saved：test_data.npz")
```
